```python
# Import required libraries
import getpass
import os

# Request OpenAI API key if not already set
if not os.environ.get("OPENAI_API_KEY"):
  os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter API key for OpenAI: ")

# Initialize the OpenAI chat model
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4o-mini", model_provider="openai")
```

```python
# Import message types from LangChain
from langchain_core.messages import HumanMessage, SystemMessage

# Define a conversation with a system instruction for translation
messages = [
    SystemMessage("Translate the following from English into Italian"),
    HumanMessage("hi!"),
]

# Invoke the model with the defined messages
model.invoke(messages)
```

```python
# Different ways to invoke the model with user input
model.invoke("Hello")  # Direct text input

# Passing input as a structured dictionary
model.invoke([{"role": "user", "content": "Hello"}])

# Using LangChain's HumanMessage wrapper
model.invoke([HumanMessage("Hello")])
```

```python
# Streaming responses from the model in real-time
for token in model.stream(messages):
    print(token.content, end="|")
```

```python
# Import ChatPromptTemplate to create structured prompts
from langchain_core.prompts import ChatPromptTemplate

# Define a system template for translation
system_template = "Translate the following from English into {language}"

# Create a structured chat prompt template with placeholders
prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_template), ("user", "{text}")]
)
```

```python
# Generate a formatted prompt by filling placeholders
prompt = prompt_template.invoke({"language": "Italian", "text": "hi!"})

# Display the formatted prompt
prompt
```

```python
# Convert the structured prompt into a list of messages
prompt.to_messages()
```

```python
# Invoke the model with the formatted prompt and print the response
response = model.invoke(prompt)
print(response.content)
```
